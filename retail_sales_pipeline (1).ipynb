```python
import pandas as pd
import random

print("--- STEP 1: GENERATING 600 ROWS OF REALISTIC RETAIL DATA ---")
cities = ['Delhi', 'Mumbai', 'Bangalore', 'Chennai', 'Hyderabad', 'Pune']
categories = ['Electronics', 'Clothing', 'Home Decor', 'Groceries', 'Footwear']
data = []

for i in range(1, 601):
    order_id = 1000 + i
    city = random.choice(cities)
    category = random.choice(categories)
    quantity = None if random.random() < 0.05 else random.randint(1, 10)
    price_per_unit = random.randint(200, 5000)
    data.append([order_id, city, category, quantity, price_per_unit])

df = pd.DataFrame(data, columns=['OrderID', 'City', 'Category', 'Quantity', 'PricePerUnit'])
print(f"Generated a dataset with {df.shape[0]} rows successfully!")

print("\n--- STEP 2: RUNNING DATA CLEANING PIPELINE ---")
df['Quantity'] = df['Quantity'].fillna(0)
print("Data Cleaning complete. Empty cells fixed.")

print("\n--- STEP 3: FEATURE ENGINEERING (REVENUE CALCULATION) ---")
df['Total_Revenue'] = df['Quantity'] * df['PricePerUnit']
print("Calculated Total_Revenue column for all 600 records.")

print("\n--- STEP 4: FINAL PORTFOLIO EXECUTIVE SUMMARY ---")
category_summary = df.groupby('Category')['Total_Revenue'].sum().reset_index()
print(category_summary.to_string(index=False))
```

```python
# 1. Save our cleaned, calculated data into a brand new CSV file
df.to_csv("cleaned_retail_sales_2026.csv", index=False)
print("📦 Success! Cleaned dataset saved locally as 'cleaned_retail_sales_2026.csv'")

print("\n📊 --- BONUS INSIGHT: TOTAL REVENUE BY CITY ---")
# 2. Let's answer your manager's second question: Which city makes the most money?
city_summary = df.groupby('City')['Total_Revenue'].sum().sort_values(ascending=False).reset_index()
print(city_summary.to_string(index=False))
```
